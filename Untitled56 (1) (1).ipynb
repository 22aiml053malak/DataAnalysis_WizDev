```python
#IMPORTING REQUIRED LIBRARY
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
#Path of CSV
path = r"C:\Users\malak\OneDrive\Desktop\Wiz_Task1.csv"
```

```python
# Load the dataset
data = pd.read_csv(path)  # Replace 'your_dataset.csv' with your actual file path

# Display the first few rows of the dataset to understand its structure
data.head()
```

```python
# Check for missing values
data.isnull().sum()
```

```python
# Segment by business scale
scale_counts = data['CScale'].value_counts()

# Plot
plt.figure(figsize=(8, 6))
sns.barplot(x=scale_counts.index, y=scale_counts.values, palette='viridis')
plt.title('Distribution of Businesses by Scale')
plt.xlabel('Business Scale')
plt.ylabel('Count')
plt.show()
```

```python
# Segment by business type
type_counts = data['Ctype'].value_counts()

# Plot
plt.figure(figsize=(10, 80))
sns.barplot(x=type_counts.values, y=type_counts.index, palette='magma')
plt.title('Distribution of Businesses by Type')
plt.xlabel('Count')
plt.ylabel('Business Type')
plt.show()
```
