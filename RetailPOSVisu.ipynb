```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file_path = r"C:\Users\malak\Downloads\retail_pos_systems_india_extended (1).csv"
```

```python
df = pd.read_csv(file_path)
```

```python
# Visualization 1: Market Share of POS Companies
plt.figure(figsize=(12, 8))
sns.barplot(x='Market Share (%)', y='Company', data=df.sort_values('Market Share (%)', ascending=False))
plt.title('Market Share of POS Companies')
plt.xlabel('Market Share (%)')
plt.ylabel('Company')
plt.show()
```

```python
# Visualization 2: Annual Revenue of POS Companies
plt.figure(figsize=(12, 8))
sns.barplot(x='Annual Revenue (₹)', y='Company', data=df.sort_values('Annual Revenue (₹)', ascending=False))
plt.title('Annual Revenue of POS Companies')
plt.xlabel('Annual Revenue (₹)')
plt.ylabel('Company')
plt.show()
```

```python
# Visualization 3: Number of Installations by Company
plt.figure(figsize=(12, 8))
sns.barplot(x='Number of Installations', y='Company', data=df.sort_values('Number of Installations', ascending=False))
plt.title('Number of Installations by Company')
plt.xlabel('Number of Installations')
plt.ylabel('Company')
plt.show()
```

```python
# Visualization 4: Distribution of Pricing Models
plt.figure(figsize=(10, 6))
sns.countplot(x='Pricing Model', data=df)
plt.title('Distribution of Pricing Models')
plt.xlabel('Pricing Model')
plt.ylabel('Count')
plt.show()
```

```python
# Visualization 5: Key Features of POS Systems
plt.figure(figsize=(10, 6))
sns.countplot(y='Key Features', data=df, order=df['Key Features'].value_counts().index)
plt.title('Key Features of POS Systems')
plt.xlabel('Count')
plt.ylabel('Key Features')
plt.show()
```

```python
# Visualization 6: Regions with Highest Usage
plt.figure(figsize=(10, 6))
sns.countplot(y='Regions with Highest Usage', data=df, order=df['Regions with Highest Usage'].value_counts().index)
plt.title('Regions with Highest Usage of POS Systems')
plt.xlabel('Count')
plt.ylabel('Regions')
plt.show()
```

```python
# Visualization 7: Correlation Matrix
plt.figure(figsize=(12, 8))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix')
plt.show()
```

```python
# Visualization 8: Market Share vs. Annual Revenue
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Market Share (%)', y='Annual Revenue (₹)', data=df, hue='Company', s=100)
plt.title('Market Share vs. Annual Revenue')
plt.xlabel('Market Share (%)')
plt.ylabel('Annual Revenue (₹)')
plt.legend(title='Company')
plt.show()
```

```python
# Visualization 9: Market Share vs. Number of Installations
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Market Share (%)', y='Number of Installations', data=df, hue='Company', s=100)
plt.title('Market Share vs. Number of Installations')
plt.xlabel('Market Share (%)')
plt.ylabel('Number of Installations')
plt.legend(title='Company')
plt.show()
```

```python
# Visualization 10: Annual Revenue vs. Number of Installations
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Annual Revenue (₹)', y='Number of Installations', data=df, hue='Company', s=100)
plt.title('Annual Revenue vs. Number of Installations')
plt.xlabel('Annual Revenue (₹)')
plt.ylabel('Number of Installations')
plt.legend(title='Company')
plt.show()
```

```python
# Visualization 11: Pricing Model by Regions with Highest Usage
plt.figure(figsize=(12, 8))
sns.countplot(x='Pricing Model', hue='Regions with Highest Usage', data=df)
plt.title('Pricing Model by Regions with Highest Usage')
plt.xlabel('Pricing Model')
plt.ylabel('Count')
plt.legend(title='Regions with Highest Usage')
plt.show()
```
